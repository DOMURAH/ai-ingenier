```python
import pandas as pd

rng = np.random.default_rng(42)

data = pd.DataFrame({
    "forfait":rng.choice(["24h","7j","30j"],200),
    "duree_session": np.round(rng.normal(120,40,200)),
    "depense" : np.round(rng.normal(5000,1500,200))
})

```

```python
sns.histplot(data=data , x="depense" , kde=True)
```

```python
sns.boxplot(data=data,x="forfait",y="depense")
```

```python
sns.scatterplot(data=data,x="duree_session",y="depense",hue="forfait")
```

```python
client = pd.DataFrame({
    "nom" : ['fifa','fita','mika'],
    "vente" : [50000,4500,12000],
    "duree" : [4,2,8]
})

sns.scatterplot(data=client,x="duree",y="vente",hue="nom")
```

```python
sns.countplot(data=client,x="vente")
```

```python
sns.heatmap(data[['duree_session','depense']].corr(),annot=True)
```

```python
datas = pd.DataFrame({
    "client": ["A", "B", "C", "D", "E", "F"],
    "duree_session": [30, 60, 90, 120, 150, 180],
    "depense": [1000, 2000, 3000, 4000, 5000, 6000]
})

sns.scatterplot(data=datas,x="duree_session",y="depense" , hue="client")
datas[['duree_session','depense']].corr()
```
